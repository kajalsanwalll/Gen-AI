```python
##Arithmetic operators

a=10
b=5
add_result = a+b
sub_result = a-b
mult_result = a*b
div_result = a/b
floor_div_result = a//b  ##floor value
modulus_result = a%b
exponent_result = a**b


print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponent_result)
```
