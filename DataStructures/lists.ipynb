```python
## accessing list elements

fruits = ["Apple", "mango", "banana","strawberry", "cherry"]
```

```python
print(fruits[3])
print(fruits[4])
print(fruits[-1]) ## last element in list
```

```python
print(fruits[1:]) ## from 1 index to end 
```

```python
print(fruits[1:3]) ## print from index 1 till 3 (not including 3)
```

```python
print(fruits)
fruits[2] = "mango"
fruits[3] = "strawberry"
print(fruits)
```

```python
## List Methods
fruits.append("orange") ## add item at the end
print(fruits)
```

```python
fruits.remove("orange") # removing the first occurance of an item
print(fruits)
```

```python
## Remove and return the last item
popped_fruit = fruits.pop()
print(popped_fruit)
```

```python
index = fruits.index("cherry")
print(index)
```

```python
fruits.insert(2,"dragonfruit")
fruits.count("dragonfruit")
print(fruits)
```

```python
print(fruits.count("dragonfruit"))
print(fruits)
```

```python
fruits.sort()
fruits
```

```python
fruits.reverse()
print(fruits)
```

```python
fruits.clear()  #removes all items from the list
print(fruits)
```
