```python
## Introduction to Tuples
#Tuples are ordered collections of items that are immutable.
#Similar to lists, but their immutability makes them different.
```

```python
##creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
## list can be converted into tuple
number = tuple([1,2,3,4,5])
number
```

```python
## tuple can be converted into a list 
list((1,2,3,4,5))
```

```python
mixed_tuple=(1, "hey sup?", 3.14, True)
print(mixed_tuple)
```

```python
## accessing tuple elements
number
```

```python
print(number[2])
print(number[-1])
```

```python
print(number[0:4])
print(number[::])
print(number[::-1])
```

```python
## Tuple Operations
number = tuple([1,2,3,4,5])
mixed_tuple=(1, "hey sup?", 3.14, True)
concatenation_tuple =  number + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple*3
```

```python
number*2
```

```python
## Immutable nature of tuples
## tuples are immutable means the elements cant be changed once assigned

lst = [1,2,3]
print(lst)

lst[1] = "kajal"
print(lst)
```

```python
print(number)
number[1] = 7
print(number)
```

```python
## Tuple methods
print(number)

print(number.count(2))
print(number.index(3)) #which index has 3 
```

```python
## Packing and unpacking tuples

#packing
packed_tuple = 1, "Hlo", 3.14
print(packed_tuple)
```

```python
#unpacking
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers = (1,2,3,4,5,6,7)

first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested tuples
## Nested lists

lst = [[1,2,3],[4,5],[6]]
lst[0][2]
lst[0][0:3]
```

```python
lst=[[1,2,3,4],[5,6],("hey", 1, 3.14)]
lst[2][0:3]
```

```python
nested_tuple = ((1,2,3),("a","b","c"), (True,False))

# access the elements inside a tuple
print(nested_tuple[2])
```

```python
## Iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item)
    print()    
```
