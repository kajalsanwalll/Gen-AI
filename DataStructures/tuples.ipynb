```python
## Introduction to Tuples
#Tuples are ordered collections of items that are immutable.
#Similar to lists, but their immutability makes them different.
```

```python
##creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
## list can be converted into tuple
number = tuple([1,2,3,4,5])
number
```

```python
## tuple can be converted into a list 
list((1,2,3,4,5))
```

```python
mixed_tuple=(1, "hey sup?", 3.14, True)
print(mixed_tuple)
```
