## Introduction to dictionaries

Dictionaries are unordered collections of items. They store data in
key-value pairs. Keys must be unique and immutable (eg: strings, numbers, tuples), while values can be of any type.

```python
## creating dictionaries
empty_dictionary = {}
print(type(empty_dictionary))
```

```python
empty_dict= dict()
empty_dict
```

```python
student = {"name": "Kajal", "age": 18, "grade": 7}
print(student)
print(type(student))
```

```python
## single key is always used
student = {"name": "Kajal", "age": 18, "name": 7}
print(student)
```

```python
## accessing dictionary elements

student = {"name": "Kajal", "age": 18, "grade": 'A'}
print(student['grade'])
print(student['age'])

## Accessing using get() method provided by dict
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name', "Not available"))
```
