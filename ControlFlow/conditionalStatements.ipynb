```python
## if statement
age = 20

if age>= 18:
    print("you're allowed to vote")
```

```python
## else statement
## it executes a block of code if the condition in the if statement is False
age = 15
if age>=18:
    print("eligible to vote")
else:
    print("not eligible to vote")    
```

```python
## elif statement
## elif statement allows you to check multiple conditions. elif = else if
age=17

if age<13:
    print("you're a child")
elif age<18:
    print("you're a teenager")
else:
    print("you're an adult")        
```

```python
## Nested conditional statements
## here you can place one or more if,else, elif statements
## inside another if, elif, else statements to create 
## nested conditional statements.

## ques = tell me if a number is even, odd or negative
num = int(input("enter a number: "))
if num >= 0:
    print("the number is positive.")
    if num%2 == 0:
        print("the number is even.")
    else:
        print("the number is odd.")
else:
    print("the number is negative.")                
```

```python
## Practical Examples

# determine if a year is leap year or not using nested conditional statements
year = int(input("enter year: "))
if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print("leap year")
        else:
            print("not a leap year") 
    else:
        print("leap year") 
else:
    print("not a leap year")                  
```

```python
## simple calculator program

num1 = float(input("enter first number: "))
num2 = float(input("enter second number: "))
operation = input("enter operator(+,-,*,/): ")

if operation == '+':
    result = num1+num2
elif operation == '-':
    result = num1-num2
elif operation == '*':
    result = num1*num2
elif operation == '/':
    if num2 != 0:
        result = num1/num2
    else:
        result = "error! division by 0"        
else:
    result = "invalid operation"
        

print(result)
```
