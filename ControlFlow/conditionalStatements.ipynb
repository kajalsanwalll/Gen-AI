```python
## if statement
age = 20

if age>= 18:
    print("you're allowed to vote")
```

```python
## else statement
## it executes a block of code if the condition in the if statement is False
age = 15
if age>=18:
    print("eligible to vote")
else:
    print("not eligible to vote")    
```

```python
## elif statement
## elif statement allows you to check multiple conditions. elif = else if
age=17

if age<13:
    print("you're a child")
elif age<18:
    print("you're a teenager")
else:
    print("you're an adult")        
```
