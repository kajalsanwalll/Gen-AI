```python
## while loop
## while loop keeps executing as long as the statement is true

count=0
while count<5:
    print(count)
    count+=1
```

```python
##loop control statements

##break
## the break statement exits the loop prematurely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue statement
## continue statement skips the current iteration and continues with next.

for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
## pass 
## the pass statement is a NULL operation; does nothing
for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
## Nested loops 
## loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i: {i},j: {j} ")
```

```python
## examples 
## calculate the sum of 1st n natural nums using a while loop and for loop

## with 'while loop' 
n = 10
sum=0
count=1
while count<=n:
    sum += count
    count = count+1

print(sum)
```

```python
## prime numbers between 1 and 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)    
```
