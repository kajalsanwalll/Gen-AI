```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
## strings
str = "Kajal Sanwal"
for i in str:
    print(i)
```

```python
## while loop
## while loop keeps executing as long as the statement is true

count=0
while count<5:
    print(count)
    count+=1
```

```python
##loop control statements

##break
## the break statement exits the loop prematurely
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## continue statement
## continue statement skips the current iteration and continues with next.

for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
## pass 
## the pass statement is a NULL operation; does nothing
for i in range(5):
    if i==3:
        pass
    print(i)
```
